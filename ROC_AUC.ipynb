```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_classification
from sklearn.metrics import roc_auc_score
from sklearn.metrics import roc_curve
```

```python
X, y = make_classification(n_samples = 10000, n_classes = 2, random_state = 42)
```

```python
from sklearn.model_selection import train_test_split 
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 42)
```

```python
dummy_model_prob = [0 for i in range(len(y_test))]
dummy_model_prob
```

```python
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression()
logreg.fit(X_train, y_train)
model_prob = logreg.predict_proba(X_test)
```

```python
model_prob = model_prob[:,1]
```

```python
model_prob
```

```python
### ROC AUC SCORES
dummy_roc_score = roc_auc_score(y_test, dummy_model_prob)
model_roc_score = roc_auc_score(y_test, model_prob)
```

```python
print('DUMMY ROC AUC SCORE ',dummy_roc_score)
print('MODEL ROC AUC SCORE ',model_roc_score)
```

```python
### AUC Curves
dummy_fpr, dummy_tpr, _ = roc_curve(y_test, dummy_model_prob)
model_fpr, model_tpr, threshold = roc_curve(y_test, model_prob)
```

```python
### PLOTS
plt.figure(figsize=(12,12))
plt.plot(dummy_fpr, dummy_tpr, linestyle = '--', c = 'r')
plt.plot(model_fpr, model_tpr, marker = '.', c = 'y')
```

```python
threshold
```

```python
fig, ax = plt.subplots(figsize=(20, 50))
ax.plot(dummy_fpr, dummy_tpr, linestyle='--', c='r', label='Dummy')
ax.plot(model_fpr, model_tpr, marker='.', c='y', label='Logistic')

for xyz in zip(model_fpr, model_tpr, threshold):
    ax.annotate('%s' % np.round(xyz[2], 2), xy=(xyz[0], xyz[1]))

ax.set_xlabel('False Positive Rate')
ax.set_ylabel('True Positive Rate')
ax.legend()
plt.show()
```
