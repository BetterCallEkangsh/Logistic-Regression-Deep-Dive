```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from  sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples= 1000, n_features= 10, n_classes=2, random_state=42)
```

```python
X
```

```python
df = pd.DataFrame(X)
df.head()
```

```python
df['Output'] = y
```

```python
df.head()
```

```python
df['Output'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logreg = LogisticRegression()
logreg.fit(X_train, y_train)
```

```python
y_pred = logreg.predict(X_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
score = accuracy_score(y_test, y_pred)
print(score)
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

### Hyperparameter Tuning and Cross Validation
GridSearchCV


```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
```

```python
model = LogisticRegression()
penalty = ['l1', 'l2', 'elasticnet']
c_values = [0.01, 0.1, 1, 10, 100]
solver = ['liblinear', 'sag', 'saga', 'newton-cg', 'lbfgs']
```

```python
params = dict(penalty = penalty, C = c_values, solver= solver)
params
```

```python
cv = StratifiedKFold(n_splits=5)
```

```python
grid = GridSearchCV(estimator=model, param_grid=params, cv=cv, scoring='accuracy', n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
scoregrid=accuracy_score(y_pred,y_test)
print(scoregrid)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```

### Randomized Search CV

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = LogisticRegression()
randomcv = RandomizedSearchCV(estimator=model, param_distributions=  params, scoring='accuracy', cv=cv)
```

```python
randomcv
```

```python
randomcv.fit(X_train, y_train)
```

```python
randomcv.best_params_
```

```python
randomcv.best_score_
```

```python
y_pred = randomcv.predict(X_test)
scorerandom=accuracy_score(y_pred,y_test)
print(scorerandom)
print(classification_report(y_pred,y_test))
print(confusion_matrix(y_pred,y_test))
```
